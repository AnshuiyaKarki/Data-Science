*Deep and shallow copy*

```python
a = np.arange(1,21)
a
```

```python
slice=a[:5]
slice=slice*10
slice #if u take a portion from the array then changes in that slice isnt refletced in orginal array->concept is called shallow copy
```

```python
a#but no change in list but changes made in list is reflected 
```

```python
a
b=a
b[0]=99 #now changes in b are also reflected in a concept is called deep copy
a
```

```python
#to create shallow copy here we use .copy 
b=a.copy()
b[1]=99
b
```

```python
a #no change in a
```
