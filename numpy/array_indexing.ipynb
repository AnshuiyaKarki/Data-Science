**NUMPY INDEXING AND SLICING OF VECTORS**

```python
arr=np.arange(11,21)
arr
```

```python
arr[6]
```

```python
arr_slice=arr[3::2]
arr_slice
```

**NUMPY INDEXING AND SLICING OF MATRIX**

```python
arr=np.arange(1,31).reshape(6,5)
arr
```

```python
arr[1]#1st element will target rows 
```

```python
arr[0,4]#0th row 4th col
```

```python
slice=arr[0:2,1:3]
slice
```

```python
arr[3:,3:]
```

```python
arr[:,2]
```
