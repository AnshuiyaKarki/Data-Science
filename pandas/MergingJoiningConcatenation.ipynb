```python
import numpy as np 
import pandas as pd
```

**Merging 2 dataframes**

```python
employees = pd.DataFrame({
    'employee_id': [1, 2, 3, 4, 5],
    'name': ['John', 'Anna', 'Peter', 'Linda', 'Bob'],
    'department': ['HR', 'IT', 'Finance', 'IT', 'HR']
})

# DataFrame 2: Salary information
salaries = pd.DataFrame({
    'employee_id': [1, 2, 3, 6, 7],
    'salary': [60000, 80000, 65000, 70000, 90000],
    'bonus': [5000, 10000, 7000, 8000, 12000]
})
```

```python
#Merging joins two DataFrames horizontally based on shared values in specific key columns (like SQL joins). 
#by default it is inner join
pd.merge(employees,salaries,on='employee_id',how='inner')
```

```python
#outer join 
pd.merge(employees,salaries,on='employee_id',how='outer')
```

```python
#left join 
pd.merge(employees,salaries,on='employee_id',how='left')
```

```python
#right join 
pd.merge(employees,salaries,on='employee_id',how='right')
```

**Concatenation of 2 dataframes**

```python
df1 = pd.DataFrame({
    'A': ['A0', 'A1', 'A2'],
    'B': ['B0', 'B1', 'B2'],
    'C': ['C0', 'C1', 'C2']
})

df2 = pd.DataFrame({
    'A': ['A3', 'A4', 'A5'],
    'B': ['B3', 'B4', 'B5'],
    'C': ['C3', 'C4', 'C5']
})
```

```python
df1
```

```python
#takes multiple DataFrames and stacks them together
#on the basis of columns here df1 data first
pd.concat([df1,df2])
```

```python
#takes multiple DataFrames and stacks them together
#on the basis of coolumns here df2 data first
pd.concat([df2,df2])
```

```python
#on the basis of rows 
pd.concat([df1,df2],axis=1)
```

**Joining of 2 dataframes**

```python
df1 = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Charlie']
}, index=[1, 2, 3])

# Second DataFrame
df2 = pd.DataFrame({
    'score': [85, 90, 75]
}, index=[2, 3, 4])
```

```python
df1
```

```python
#combines two DataFrames along their indexes (row labels).
df1.join(df2) 
```

```python
df2.join(df1)
```

```python
df2.join(df1,how='outer')
```
