# Pandas Series 

A Series is a one dimensional labeled array capable of holding any data type. The axis labels are collectively called the index. 
It is like array except array is horizontal and has indexes, series is vertical and has labels.

## Creating Series

There are multiple ways to create a pandas Series:

```python
import numpy as np
import pandas as pd
```

```python
labels=['a','b','c']
my_list=['1','2','3']
arr= np.array([10,20,30])
d={1:100,2:200,3:300}
```

```python
pd.Series(my_list)
```

```python
pd.Series(arr,labels) #if u pass labels it gives custom labels
```

```python
pd.Series(d) #when u pass dictionary in series it assigns keys as  automatic labels
```
