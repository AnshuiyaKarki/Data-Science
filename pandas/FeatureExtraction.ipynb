```python
import numpy as np 
import pandas as pd 
```

```python
#loading the data 
df=pd.read_csv(r'anime.csv')
```

```python
df.head()
```

```python
df.loc[4]['Title']
```

```python
def extract_episodes(txt): 
    check = False
    data=""
    for i in txt:
        if i==')':
            check=False 
            continue 
        if i=='(': 
            check=True
            continue 
            
        if check==True:
            data+=i
    return data 
```

```python
df['Episodes']=df['Title'].apply(extract_episodes)
df.head()
```

```python
df['Episodes']=df['Episodes'].str.replace("eps","")
df.head()
```

```python
df.loc[0]['Episodes']
```

```python
df['Episodes']=df['Episodes'].astype(int)
df.head()
```

```python
#make a new column for episode count->done
#make a new column for time stamp ->done
#which anime has the highest score ->done
#give me top 5 highest scoring anime ->done
#which anime has the highest episode count ->done
#animes with top 5 episode count 
#which is the longest running anime 
```

```python
df.loc[2]['Title']
```

```python
def extract_timestamp(txt):
    for i in range(len(txt)): 
        if txt[i]==')':
            return txt[i+1:i+20]
```

```python
df['TimeStamp']=df['Title'].apply(extract_timestamp)
df.head()
```

```python
#which anime has the highest score 
df[df['Score']==df['Score'].max()]
```

```python
#give me top 5 highest scoring anime
df.sort_values(by='Score',inplace=True,ascending=False)
df.head()
```

```python
#which anime has the highest episode count 
df[df['Episodes']==df['Episodes'].max()]
```

```python
#animes with top 5 episode count
df.sort_values(by='Episodes',ascending=False).head()
```

```python
df[['Start_Text','End_Text']]=df['TimeStamp'].str.split(' - ',expand=True)
df.head()
```

```python
df['Start']=pd.to_datetime(df['Start_Text'],format='%b %Y', errors='coerce')
df['End']=pd.to_datetime(df['End_Text'],format= '%b %Y', errors='coerce')
```

```python
df.head()
```

```python
df['Duration']=(df['End']-df['Start']).dt.days
df.sort_values(by='Duration',ascending=False).head(1)
```
