When you combine multiple series together it is called Dataframe

```python
import numpy as np 
import pandas as pd
```

**Creating a Dataframe**

```python
#using dictionary 
data={
    'Name':['Ram','Sita','Jignu','Anshuiya'],
    'Age':[21,24,25,23],
    'Country':['India','Italy','USA','Nepal'],
    'Salary':['1lakh','2lakh','3lakh','10crore']
}

pd.DataFrame(data)
```

```python
#using list
data_list=[
    ['Ram',21,'India','1lakh'],
    ['Sita',24,'Italy','2lakh'],
    ['Jignu',25,'USA','3lakh'],
    ['Anshuiya',23,'Nepal','10crore']
]
pd.DataFrame(data_list)

#in this case indexes 0 ,1 ,2 come as labels 
```

```python
#using list u have to manually customize labels
data_list=[
    ['Ram',21,'India','1lakh'],
    ['Sita',24,'Italy','2lakh'],
    ['Jignu',25,'USA','3lakh'],
    ['Anshuiya',23,'Nepal','10crore']
]

column_name=['Name','Age','Country','Salary']
df2=pd.DataFrame(data_list,columns=column_name)
```

**Selection and Indexing of Columns**

```python
df2['Name']  #have to pass labels 
```

```python
df2[['Name','Country']] #for multiple have to past a list of lists
```

**Creating a new column**

```python
df2['Designation']=['Doctor','Lawyer','Doctor','Software Development Engineer']
df2
```

**Removing columns**

```python
df2.drop('Designation',axis=1) #vertical axis =1 but thisis only dropped in copy in real data changes arent reflected
```

```python
df2 #changes not reflected
```

```python
df2.drop('Designation',axis=1, inplace = True)
```

```python
df2
```

```python
#to drop multiple columns drop list 
df2.drop(['Name','Country'],axis=1)
```

**Selecting Rows**

```python
df2.loc[3] #loc means location
```

```python
df2.loc[[1,3]]
```

```python
df2.iloc[3] #similar to loc iloc is index loc
```

**Selecting Subsets of Rows and Columns**

```python
df2
```

```python
df2.loc[[0,1]][['Country','Salary']]
```

```python
df2.loc[[2,3]][['Name','Age']]
```

**Conditional Selection**

```python
#I only see those people whose age is above 21 
```

```python
df2[df2['Age']>21]
```

```python
#I only want those people whose age is above 21 and are from Nepal cause they are extremely talented
```

```python
df2[(df2['Age']>21) & (df2['Country']=='Nepal')]
```
