```python
import numpy as np 
import pandas as pd
```

**Finding Missing Data**

```python
data={
    'A':[1,2,np.nan,4,5],
    'B':[np.nan,2,3,4,5],
    'C':[1,2,3,np.nan,np.nan],
    'D':[1,np.nan,np.nan,np.nan,5]
}
df=pd.DataFrame(data)
```

```python
df
```

```python
#to see where if nan exists 
df.isna()
```

```python
#to count the no of nan in each column
df.isna().sum()
```

```python
#if any missing value in each column->sab ma null values cha ki kunai na kunai ma chaina
df.isna().any()
```

**Removing Missing Data**

```python
df
```

```python
#to drop all ROWS with nan (not a number )values 
df.dropna() 
```

```python
#to keep rows with atleast thresh NON NAN values 
df.dropna(thresh=3)
```

**Filling the missing Data**

```python
#to fill COLS the nan values with 0s
df.fillna(0)
```

```python
#to fill COLS nan with custom made values
values={'A':100, 'B':200,'C':300,'D':400}
df.fillna(value=values)
```

```python
#to fill nan with values with df mean 
df.fillna(df.mean())
```
